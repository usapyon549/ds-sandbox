```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import polars as pl
import seaborn as sns
from datasets import load_dataset
```

## 1. データの読み込み

```python
dataset = load_dataset("stanfordnlp/imdb")
```

```python
dataset["train"]
```

```python
pl_df_train = dataset["train"].to_polars()
```

```python
pl_df_train.head(5)
```

## 2. データの中身の確認

```python
pl_df_train.null_count().sum()
# →欠損なし
```

### 2.1 目的変数の分布の確認

```python
pl_df_train["label"].value_counts()
```

```python
# 目的変数の分布の確認

data = pl_df_train["label"].value_counts()
sns.barplot(data=data,
           x="label",
            y= "count"
           )
plt.xlabel("label")
plt.ylabel("Frequency")
plt.title("Barplot of label")
plt.show()
```

### 2.3　テキストの文字長さの分布の確認

```python
sns.histplot(data=pl_df_train["text"].str.len_chars().to_frame(), x="text")

plt.xlabel("text_len")
plt.ylabel("Frequency")
plt.title("Histogram of text_len")
plt.show()

# →文字の長さがマイナスなどの異常値はなさそう
```
