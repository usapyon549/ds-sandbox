### 3.2 サイクルごとの推移（移動平均)

```python
# バッテリーごとに電流のグラフ

window = 5

fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(10, 8))

for i, battery_id in enumerate(list_battery_ids):

    ax = axes[i]
    df = df_eda[df_eda["battery_id"] == battery_id]    

    ax.plot(df["cycle"], df["chI"].rolling(window=window).mean(), label="SMA:charge I")
    ax.plot(df["cycle"], df["disI"].rolling(window=window).mean(), label="SMA:discharge I")

    axes[-1].set_xlabel("cycle")
    ax.set_ylabel("current")
    ax.set_title(f"SMA of Current:Battery_{battery_id}")
    ax.legend(loc="upper right") # グラフごとに凡例を表示

# 一番下のグラフだけにx軸ラベルをつける（sharex=Trueにしているため）
# axes[-1].set_xlabel("cycle")

# グラフ同士の文字が重ならないように自動調整
plt.tight_layout()
plt.show()
```

### 3.3 サイクルごとの推移（傾き）

```python
# 傾きを計算する関数
# rolling(window)内で
# 一次近似の傾きを返す

def calc_slope(y):

    list_x = np.arange(len(y))
                       
    slope, _ = np.polyfit(list_x, y, 1)

    return slope
```

```python
# バッテリーごとに電流のグラフ
window = 5

fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(10, 8))

for i, battery_id in enumerate(list_battery_ids):

    ax = axes[i]
    df = df_eda[df_eda["battery_id"] == battery_id]    

    ax.plot(df["cycle"], df["chI"].rolling(window=window).apply(calc_slope), label="Slope:charge I")
    ax.plot(df["cycle"], df["disI"].rolling(window=window).apply(calc_slope), label="Slope:discharge I")

    axes[-1].set_xlabel("cycle")
    ax.set_ylabel("current")
    ax.set_title(f"Slope of Current:Battery_{battery_id}")
    ax.legend(loc="upper right") # グラフごとに凡例を表示

# 一番下のグラフだけにx軸ラベルをつける（sharex=Trueにしているため）
# axes[-1].set_xlabel("cycle")

# グラフ同士の文字が重ならないように自動調整
plt.tight_layout()
plt.show()
```

```python
# バッテリーごとに電圧のグラフ
window = 5

fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(10, 8))

for i, battery_id in enumerate(list_battery_ids):

    ax = axes[i]
    df = df_eda[df_eda["battery_id"] == battery_id]    

    ax.plot(df["cycle"], df["chV"].rolling(window=window).apply(calc_slope), label="Slope:charge V")
    ax.plot(df["cycle"], df["disV"].rolling(window=window).apply(calc_slope), label="Slope:discharge V")

    axes[-1].set_xlabel("cycle")
    ax.set_ylabel("voltage")
    ax.set_title(f"Slope of Voltage:Battery_{battery_id}")
    ax.legend(loc="upper right") # グラフごとに凡例を表示

# 一番下のグラフだけにx軸ラベルをつける（sharex=Trueにしているため）
# axes[-1].set_xlabel("cycle")

# グラフ同士の文字が重ならないように自動調整
plt.tight_layout()
plt.show()
```
