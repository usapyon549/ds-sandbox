## 3. 数値型データの描画

```python
dict_dtype_map["num"]
```

```python
def draw_histo(df, col):

    sns.histplot(data=df, x=col)
    plt.ylabel("frequency")
    plt.title(f"Histogram of {col}")
    plt.show()
```

```python
draw_histo(df=df_data, col="age")
```

```python
draw_histo(df=df_data, col="education-num")
```

```python
draw_histo(df=df_data, col="capital-gain")

# 外れ値あり。学習の時は前処理を入れる。
# 0が大半を占め、一部だけ大きな値を持つ右裾の長い分布
```

```python
draw_histo(df=df_data, col="capital-loss")

# 外れ値あり。学習の時は前処理を入れる。
# 0が大半を占め、一部だけ大きな値を持つ右裾の長い分布
```
